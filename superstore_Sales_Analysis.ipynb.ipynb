```python
df['Order Date'] = pd.to_datetime(df['Order Date'])

df['Ship Date'] = pd.to_datetime(df['Ship Date'])
```

```python
# Sales by Category

df.groupby("Category")["Sales"].sum()
```

```python
# 01.Sales by Category - Bar Chart

plt.figure(figsize =(7,5))
plt.bar(df.groupby("Category")["Sales"].sum().index,df.groupby("Category")["Sales"].sum(), color = "brown")
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.show()
```

```python
# 02.Monthly Sales Trend - line chart

df["Month"] = df["Order Date"].dt.month

monthly_sales = df.groupby("Month")["Sales"].sum()

plt.figure(figsize=(7,4))
plt.plot(monthly_sales.index, monthly_sales.values,marker = "o")
plt.title("Monthly sales trend")
plt.xlabel("Monthly Sales Trend ")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
# 03.Top Products - bar chart

top_products = df.groupby("Product Name")["Sales"].sum().nlargest(10)

plt.figure(figsize = (10,6))
plt.barh(top_products.index,top_products.values,color = "red")
plt.title("Top 10 Productd by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Production")
plt.show()
```

```python
# 04.Region Sales - pie chart

region_sales = df.groupby("Region")["Sales"].sum()

plt.figure(figsize = (7,4))
plt.pie(region_sales.values,labels=region_sales.index,autopct="%1.1f%%",startangle=90,wedgeprops={"width":0.4},pctdistance=0.8)
plt.title("Region Sales Distribution")
plt.show()
```

```python
# 05.Region Sales - Donut Chart

region_sales = df.groupby("Region")["Sales"].sum()

plt.figure(figsize = (7,4))
plt.pie(region_sales.values,labels=region_sales.index,autopct="%1.1f%%")
plt.title("Region Sales Distribution")
plt.show()
```

```python
# 06.Discount vs Profit - Scatter plot

plt.figure(figsize=(7,4))
plt.scatter(df["Discount"],df["Profit"], alpha=0.4)
plt.xlabel("Discount")
plt.ylabel("Profit")
plt.title("Discount vs Profit")
plt.show()
```

```python
# 07.Sales Distribution - Histogram

plt.figure(figsize=(7,4))
plt.hist(df["Sales"],bins=10,edgecolor="black")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.title("Sales Distribution")
plt.show()
```

```python
# 08.Sales Data - Correlation heatmap

plt.figure(figsize=(7,4))
corr = df[["Sales", "Profit","Discount","Quantity"]].corr()

sns.heatmap(corr, annot=True,cmap="coolwarm", fmt=".2f")
plt.title("Sales Data - Correlation Heatmap")
plt.show()
```

```python
# 09.Quantity Disturbution - Box plot

plt.figure(figsize=(7,4))
sns.boxplot(x=df["Quantity"])

plt.title("Quantity Distribution")
plt.xlabel("Quantity")
plt.show()
```

```python
# 10.Sales by Ship Mode - Pie Chart

ship_sales = df.groupby("Ship Mode")["Sales"].sum()

plt.figure(figsize=(7,4))
plt.pie(ship_sales.values,labels=ship_sales.index,autopct="%1.1f%%",startangle=90)
plt.title("Sales by Ship Mode")
plt.show()

```

```python
# 11.Sales vs Profit - Scatter Plot

plt.figure(figsize=(7,4))
sns.scatterplot(data=df,x="Sales",y="Profit",hue="Category",alpha=0.6)

plt.title("Sales vs Profit by Category")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.show()
```
